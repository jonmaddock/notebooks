# Active constraints in EU-DEMO models

Take the current large tokamak input file from Process, convert it to inequalities. Done using `dakota-process/design_points/prep_sol_for_uq.ipynb`. Then optimise.

In solution, extract inequality values: which constraints are active?




```python
# (Using dakota env)
import pandas as pd
import re
```

```python
with open("large_tokamak_ineqs_MFILE.DAT") as f:
    lines = f.readlines()

cons = []
for line in lines:
    match = re.match(r"\s*(\S*)\s*\(ineq_con(\d{3}).*(-?\d\.\d*E[+-]\d*)", line)
    if match:
        cons.append(
            [
                match.group(1).replace("_", " ").strip(),
                int(match.group(2)),
                float(match.group(3)),
            ]
        )

df = pd.DataFrame(data=cons, columns=["Inequality constraint", "Number", "Value"])
df
```

```python
sorted_df = df.sort_values(by="Value")
```

```python
sorted_df.to_latex("cons_table.tex", float_format="%.3e", index=False, longtable=True)
```
