```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
rng = np.random.default_rng()
N = 10000
BINS = 40
samples = rng.normal(loc=1.0e-5, scale=1.0e-4, size=N)
plt.hist(samples, bins=BINS)
```

```python
rng = np.random.default_rng()
N = 10000
BINS = 40
samples = rng.normal(loc=1.0e-1, scale=1.0e-3, size=N)
plt.hist(samples, bins=BINS)
```

```python
samples = rng.lognormal(size=N)
plt.hist(samples, bins=BINS)
```

```python
samples = rng.lognormal(mean=1.0e-1, sigma=1.5e0, size=N * 50)
plt.hist(samples, bins=BINS * 50)
plt.xlim(0.0, 1.0)
```

```python
samples = rng.lognormal(mean=0.0, sigma=5.0e-1, size=N * 50)
plt.hist(samples, bins=BINS * 50)
# plt.xlim(0.0,5.0)
```

## Mode of lognormal

```python
mode = 1e-8
np.log(mode)
```

```python
mu = 0
sigma = np.sqrt(-(np.log(mode) - mu))
sigma
```

```python
samples = rng.lognormal(mean=0.0, sigma=4.29, size=N * 50)
samples_filtered = samples[samples < 1.0e-3]
plt.hist(samples_filtered, bins=BINS * 50)
plt.xlim(0.0, 1.0e-3)
```
