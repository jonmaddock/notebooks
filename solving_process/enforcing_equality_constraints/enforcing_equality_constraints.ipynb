## Try to solve for equality constraint
`try_soln` has high W and is going to try and solve only the power balance constraint with only te as the parameter, using scipy's `fsolve()`.


```python
single_run = SingleRun("large_tokamak_sol_high_W_try_soln_IN.DAT")
single_run.run()
```

```python
con2, _, _, _, _ = process.fortran.constraints.constraint_eqn_002()
con2
```

It solves! Con2 is satisfied again. What does the solution look like? Extract solution from consistency solve, overwrite into high W input file and re-run.

```python
ineqs_in_name = "large_tokamak_sol_high_W_IN.DAT"
mfile_name = "large_tokamak_sol_high_W_try_soln_MFILE.DAT"
sol_in_name = "large_tokamak_sol_high_W_try_soln_sol_IN.DAT"

convert(
    original_in_name=ineqs_in_name,
    mfile_name=mfile_name,
    new_in_name=sol_in_name,
    no_optimisation=True,
    n_equalities=3,
    remove_f_value_inits=False,
)
```

Now re-run the once-through high W input file, at the consistency solve point.

```python
single_run = SingleRun("large_tokamak_sol_high_W_try_soln_sol_IN.DAT")
single_run.run()
```

```python
con2, _, _, _, _ = process.fortran.constraints.constraint_eqn_002()
con2
```

The global power balance equality constraint has the same value as before: it's satisfied. Zoom in to see equalities only.

```python
runs_metadata = [
    RunMetadata("large_tokamak_sol_MFILE.DAT", "LT"),
    RunMetadata("large_tokamak_sol_high_W_MFILE.DAT", "LT high W"),
    RunMetadata(
        "large_tokamak_sol_high_W_try_soln_sol_MFILE.DAT", "LT high W consistency soln"
    ),
]
fig, once_through_df = plot_mfile_solutions_constraints(
    runs_metadata=runs_metadata,
    title="Equality constraint values at large tokamak solution",
    eqs_only=True,
)
```
