# Use scipy's fsolve to solve some equalities

```python
from scipy.optimize import fsolve
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

2 equality constraints (equations), taking 2 parameters each. Plot each over x,y plane.

```python
def func(x):
    return [x[0] * np.cos(x[1]) - 4, x[1] * x[0] - x[1] - 5]


x = y = np.linspace(0, 7, 100)
X, Y = np.meshgrid(x, y)
Z = func([X, Y])
fig, ax = plt.subplots(ncols=2)
fig.set_figwidth(10)
# create a single norm to be shared across all images
norm = matplotlib.colors.Normalize(vmin=np.min(Z), vmax=np.max(Z))
ax[0].contourf(X, Y, Z[0], norm=norm)
ax[0].set_title("Value of constraint 1")
contour_set = ax[1].contourf(X, Y, Z[1], norm=norm)
ax[1].set_title("Value of constraint 2")
cbar = fig.colorbar(contour_set, ax=ax)
ax[0].set_xlabel("x")
ax[0].set_ylabel("y")
ax[1].set_xlabel("x")
ax[1].set_ylabel("y")

# Line for 0
levels = [0]
ax[0].contour(X, Y, Z[0], levels, colors="red")
ax[1].contour(X, Y, Z[1], levels, colors="red")

# Solve and plot solution
root = fsolve(func, [1, 1])
ax[0].scatter(root[0], root[1], c="r")
ax[1].scatter(root[0], root[1], c="r")
fig.suptitle("Values of 2 equality constraint equations and their solution")
```

Plot solutions of each equation in blue and red, with solution as red circle.

```python
fig, ax = plt.subplots()
levels = [0]
ax.contour(X, Y, Z[0], levels, colors="blue")
ax.contour(X, Y, Z[1], levels, colors="red")
ax.scatter(root[0], root[1], c="r")
ax.set_xlabel("x")
ax.set_ylabel("y")
```
