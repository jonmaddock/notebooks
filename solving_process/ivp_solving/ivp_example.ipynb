```python
from scipy.integrate import solve_ivp
import numpy as np
import matplotlib.pyplot as plt
```

```python
def exp_decay(t, y):
    return -0.5 * y


initial_values = [2, 4, 8]
time_span = [0, 10]
sol = solve_ivp(fun=exp_decay, t_span=time_span, y0=initial_values)
fig, ax = plt.subplots()
for i in range(len(initial_values)):
    ax.plot(sol.t, sol.y[i, :])
```

## Solve ODE in 2 parameters
Make something up.

```python
def func(t, y):
    df0dt = 10 - 0.5 * y[0] - y[1]
    df1dt = y[1] + 2 * (y[0] + df0dt)
    return [df0dt, df1dt]


time_span = [0, 10]
initial_values = [100, -50]
sol = solve_ivp(fun=func, t_span=time_span, y0=initial_values)
# Plot
fig, ax = plt.subplots()
for i in range(len(initial_values)):
    ax.plot(sol.t, sol.y[i, :], label=f"y[{i}]")
ax.legend()
```
